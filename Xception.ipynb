#Xception Keras Pretrained Model

***IMPORTS***

```python
import tensorflow as tf
from tensorflow.keras.applications import Xception
from tensorflow.keras.applications.xception import preprocess_input,decode_predictions
from tensorflow.keras.preprocessing.image import load_img,img_to_array
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

***Model Loading***

```python
# Load the pretrained Xception Model
model = Xception(weights='imagenet')
```

```python
model.summary()
```

```python
# Prediction function
def predict_img(model,img_pth):
  img = load_img(img_pth,target_size=(299,299))
  img_ary = img_to_array(img)
  img_ary = np.expand_dims(img_ary,axis=0)
  img_ary = preprocess_input(img_ary)

  preds = model.predict(img_ary)
  return decode_predictions(preds,top=3)[0]
```

```python
import requests

def download_image(url,save_pth):
  """
  Downloads an image from a given URL and saves it to a specified path.

  Args:
      url: The URL of the image.
      save_path: The path where the image will be saved.
  """
  try:
    response = requests.get(url,stream=True)
    response.raise_for_status()
    with open(save_pth,'wb') as f:
      for chunk in response.iter_content(chunk_size=8192):
        f.write(chunk)
    print(f"Image download successfully: {save_pth}")
  except requests.exceptions.RequestException as e:
    print(f"Error download during: {e}")
```

```python
download_image('https://cdn.mos.cms.futurecdn.net/HjFE8NKWuCmgfHCcndJ3rK.jpg','zebra.jpg')
```

```python
# Example usage
img_pth = '/content/zebra.jpg'

# Load and preprocess the image for prediction
img = load_img(img_pth, target_size=(299, 299))
preds = predict_img(model,img_pth)


# Display top prediction
print("TOP PREDICTIONS: ")
for cls_id,description,score in preds:
  print(f"{cls_id}. {description}: {score:.4f}")


# Display the image
plt.imshow(load_img(img_pth,target_size=(299,299)))
plt.axis("off")
plt.title("INPUT IMAGE")
plt.show()
```
